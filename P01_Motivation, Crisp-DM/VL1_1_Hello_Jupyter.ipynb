## 3. tabellarische Daten als pandas-DataFrame

Tabellarische Daten werden in `pandas` typischerweise als **DataFrame** gespeichert.

- Eine **Zeile** beschreibt eine Beobachtung bzw. statistische Einheit.
- Eine **Spalte** beschreibt ein Merkmal.
- Eine einzelne DataFrame-Spalte ist eine **Series**.

Für die folgenden Beispiele wird ein kleiner Kundendatensatz direkt im Notebook erzeugt.


````python
data = {
    "Kundennr": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "Alter": [25, 42, 37, 51, 29, 46, 33, 40],
    "Einkommen": [45, 95, 72, 120, 58, 88, 64, 105],
    "Region": ["Nord", "West", "Nord", "Süd", "West", "Süd", "Nord", "West"],
    "Umsatz": [1.8, 4.7, 3.1, 6.2, 2.4, 4.1, np.nan, 5.3],
    "Online": [True, True, False, True, False, True, True, False],
}

df = pd.DataFrame(data)
df
````

### Series oder DataFrame?

Eine einzelne Spalte mit einfachen eckigen Klammern liefert eine `Series`.  
Mehrere Spalten werden mit einer **Liste von Spaltennamen** ausgewählt und ergeben wieder einen `DataFrame`.


````python
print(type(df["Einkommen"]))
print(type(df[["Einkommen"]]))
````

## 4. Einen DataFrame kennenlernen
### Verbindung zur Statistik

Tabellarische Daten lassen sich direkt mit den aus der Statistik bekannten Begriffen verbinden:

| Statistik | `pandas` / DataFrame |
|---|---|
| statistische Einheit / Beobachtung | Zeile |
| Merkmal / Variable | Spalte |
| Merkmalswert | Zellenwert |

Ein `DataFrame` stellt damit einen tabellarischen Datensatz dar:

- **Zeilen** beschreiben die einzelnen Beobachtungen bzw. statistischen Einheiten,
- **Spalten** beschreiben die Merkmale bzw. Variablen,
- **Zellen** enthalten die beobachteten Merkmalswerte.

Eine einzelne Spalte eines `DataFrame` ist in `pandas` eine **`Series`**.

Typische erste Befehle sind:

- `head()` – erste Zeilen,
- `tail()` – letzte Zeilen,
- `shape` – Anzahl Zeilen und Spalten,
- `columns` – Spaltennamen,
- `dtypes` – technische Datentypen,
- `info()` – kompakter Überblick einschließlich nicht-fehlender Werte.

## 5. Spalten und Zeilen auswählen

### Spaltenauswahl

## 6. Daten mit Bedingungen filtern

Sehr häufig werden nur Beobachtungen benötigt, die bestimmte Kriterien erfüllen.

Beispiel: Kunden mit einem Einkommen von mindestens 80.


````python
# Filtere alle Zeilen, bei denen das Einkommen mindestens 80 beträgt
df[df["Einkommen"] >= 80]
````

Mehrere Bedingungen werden mit `&` (und) bzw. `|` (oder) verknüpft. Einzelne Bedingungen stehen dabei in Klammern.


````python
# Filtere alle Kunden mit einem Einkommen von mindestens 80, die Online-Banking nutzen
df[
    (df["Einkommen"] >= 80)
    & (df["Online"])
]
````

Mit `loc` lassen sich Filterung und Spaltenauswahl direkt kombinieren.


````python
# Wähle alle Kunden aus der Region West und zeige Kundennr, Einkommen und Umsatz
df.loc[
    df["Region"] == "West",
    ["Kundennr", "Einkommen", "Umsatz"]
]
````

## 7. DataFrames gezielt verändern

Bei Analysen empfiehlt es sich häufig, vor größeren Veränderungen eine Kopie anzulegen. Dadurch bleibt der ursprüngliche DataFrame erhalten.


````python
# Erstelle eine unabhängige Kopie von df mit dem Namen df_work
df_work = df.copy()
````

### Neue Spalte erzeugen

Eine neue Variable kann aus vorhandenen Merkmalen abgeleitet werden. `np.where()` ist für einfache Fallunterscheidungen praktisch.


````python
# Erzeuge die Spalte Einkommensgruppe: hoch bei Einkommen ab 80, sonst niedrig, und zeige die relevanten Spalten
df_work["Einkommensgruppe"] = np.where(
    df_work["Einkommen"] >= 80,
    "hoch",
    "niedrig"
)

df_work[["Kundennr", "Einkommen", "Einkommensgruppe"]]
````

### Werte gezielt ändern

Für bedingte Zuweisungen ist `loc` gut geeignet.


````python
# Erzeuge die Spalte Premium und setze sie auf True für Kunden mit Einkommen ab 90 und Online-Banking
df_work["Premium"] = False

df_work.loc[
    (df_work["Einkommen"] >= 90) & (df_work["Online"]),
    "Premium"
] = True

df_work[["Kundennr", "Einkommen", "Online", "Premium"]]
````

### Spalten umbenennen oder entfernen

`rename()` verändert Namen, `drop()` entfernt Zeilen oder Spalten. Für Demonstrationszwecke wird jeweils eine Kopie erzeugt.   
Da inplace=False standardmäßig gilt, wird ein neues DataFrame erzeugt und df_work bleibt unverändert.

````python
# Benenne die Spalte Umsatz in Monatsumsatz um.
df_renamed = df_work.rename(columns={"Umsatz": "Monatsumsatz"})
df_renamed.head(2)
````

````python
# Entferne die Spalte Premium und zeige die ersten zwei Zeilen
df_reduced = df_work.drop(columns=["Premium"])
df_reduced.head(2)
````

## 8. Fehlende Werte erkennen und behandeln

Fehlende Werte werden in `pandas` häufig als `NaN` dargestellt.

Zunächst sollte festgestellt werden, **ob und wo Werte fehlen**.


````python
# Zeige für jede Spalte die Anzahl der fehlenden Werte
df.isna().sum()
````

Für eine technische Bearbeitung stehen beispielsweise `dropna()` und `fillna()` zur Verfügung.

Welche Behandlung fachlich sinnvoll ist, hängt jedoch vom Datensatz und von der Fragestellung ab. In der Vorlesung wird diese Entscheidung im Zusammenhang mit **Data Understanding** und **Data Preparation** behandelt.


````python
# Entferne alle Zeilen mit fehlenden Werten und gib die Zeilenanzahl vorher und nachher aus
df_without_missing = df.dropna()

print("Zeilen vorher:", len(df))
print("Zeilen nach dropna():", len(df_without_missing))
````

````python
# Fülle fehlende Werte in Umsatz mit dem Median der Spalte und zeige den DataFrame
df_filled = df.copy()

median_umsatz = df_filled["Umsatz"].median()
df_filled["Umsatz"] = df_filled["Umsatz"].fillna(median_umsatz)

df_filled
````

## 10. Gruppenauswertungen mit `groupby`

Mit `groupby()` werden Beobachtungen nach Kategorien gruppiert. Anschließend können Kennzahlen je Gruppe berechnet werden.


````python
# Berechne das mittlere Einkommen für jede Region
df.groupby("Region")["Einkommen"].mean()
````

````python
# Gruppiere nach Region und berechne Anzahl der Kunden, mittleres Einkommen und mittleren Umsatz
df.groupby("Region").agg(
    Anzahl=("Kundennr", "count"),
    Mittleres_Einkommen=("Einkommen", "mean"),
    Mittlerer_Umsatz=("Umsatz", "mean"),
)
````

## 11. Einfache Visualisierungen

Grafiken unterstützen dabei, Verteilungen und Zusammenhänge schnell zu erkennen.

### Balkendiagramm für Kategorien


````python
# Erstelle ein Balkendiagramm mit der Anzahl der Kunden je Region und beschrifte Achsen und Titel
df["Region"].value_counts().plot(kind="bar")

plt.xlabel("Region")
plt.ylabel("Anzahl")
plt.title("Kunden nach Region")
plt.tight_layout()
plt.show()
````

### Achtung bei Python-Histogrammen

Python-Funktionen zur Erstellung eines Histogramms zeigen standardmäßig häufig die **Anzahl der Beobachtungen pro Klasse** auf der y-Achse.

In der in der Vorlesung verwendeten statistischen Terminologie wird das Histogramm als **Dichtediagramm** dargestellt:

- Die Werte werden zu Klassen zusammengefasst.
- Entscheidend ist die **Fläche der Balken**.
- Die Fläche eines Balkens entspricht der **relativen Häufigkeit** in der jeweiligen Klasse.
- Die Gesamtfläche des Histogramms beträgt **1**.

Bei gleich breiten Klassen unterscheiden sich Häufigkeits- und Dichtedarstellung nur durch die Skalierung der y-Achse. Besonders wichtig ist die Flächeninterpretation bei **unterschiedlichen Klassenbreiten**


````python
# Erstelle ein Histogramm als Dichtediagramm
df["Einkommen"].plot(
    kind="hist",
    bins=5,
    density=True,
    edgecolor="black"
)

plt.xlabel("Einkommen")
plt.ylabel("Dichte")
plt.title("Verteilung des Einkommens")
plt.tight_layout()
plt.show()
````

### Scatterplot für zwei metrische Variablen


````python
# Erstelle ein Streudiagramm von Einkommen und Umsatz und beschrifte Achsen und Titel
df.plot.scatter(x="Einkommen", y="Umsatz")

plt.xlabel("Einkommen")
plt.ylabel("Umsatz")
plt.title("Einkommen und Umsatz")
plt.tight_layout()
plt.show()
````

### Seaborn: zusätzliche Informationen direkt in der Grafik

`Seaborn` baut auf Matplotlib auf und eignet sich besonders für die explorative Datenanalyse. Spalten eines DataFrames können direkt als Achsen oder zusätzliche Gruppierungsmerkmale verwendet werden.

Im folgenden Beispiel zeigt die Punktfarbe zusätzlich die **Region**. Dadurch lässt sich in einer einzigen Grafik prüfen, ob sich neben dem Zusammenhang zwischen Einkommen und Umsatz auch Unterschiede zwischen Gruppen andeuten.


````python
# Erstelle mit Seaborn ein Streudiagramm von Einkommen und Umsatz und unterscheide die Regionen durch die Farbe
sns.scatterplot(
    data=df,
    x="Einkommen",
    y="Umsatz",
    hue="Region"
)

plt.title("Einkommen und Umsatz nach Region")
plt.tight_layout()
plt.show()
````

## 12. CSV-Dateien einlesen

Reale Datensätze werden häufig als CSV-Dateien bereitgestellt.

Der Standardfall lautet:


Je nach Datei können zusätzliche Parameter nötig sein, zum Beispiel:

```python
pd.read_csv(
    "datei.csv",
    sep=";",
    decimal=",",
    encoding="utf-8"
)
```

Für die Vorbereitung wird das Einlesen ohne zusätzliche Datei demonstriert.


````python
# Lies die folgenden semikolongetrennten CSV-Daten aus dem String csv_text in einen DataFrame df_csv ein
from io import StringIO

csv_text = """Produkt;Preis;Menge
A;12.50;3
B;8.90;5
C;15.20;2
"""

df_csv = pd.read_csv(
    StringIO(csv_text),
    sep=";"
)

df_csv
````
