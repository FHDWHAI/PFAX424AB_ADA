## 5. Technischer Datentyp und fachliches Skalenniveau

Die technische Speicherung in Python und die fachliche Bedeutung einer Variable sind zu unterscheiden. Für die Auswahl geeigneter Kennzahlen und Darstellungen ist das **fachliche Skalenniveau** entscheidend.


```python
# Numerisch und nicht numerisch gespeicherte Variablen
numeric_columns = df_bank.select_dtypes(include="number").columns.tolist()
categorical_columns = df_bank.select_dtypes(exclude="number").columns.tolist()

print("Numerisch gespeichert:", numeric_columns)
print("Nicht numerisch gespeichert:", categorical_columns)
```

### Technischer Datentyp und fachliches Skalenniveau

Der Python-Datentyp beschreibt die **technische Speicherung**; das Skalenniveau beschreibt die **fachliche Bedeutung**.

Beispiel: `Education` ist technisch als Ganzzahl gespeichert. Die Werte 1, 2 und 3 stehen jedoch für **geordnete Kategorien**.

Für die Auswahl geeigneter Kennzahlen und Darstellungen ist daher das fachliche Skalenniveau entscheidend:

- metrisch: z. B. `Age`, `Income`, `CCAvg`
- kategorial nominal: z. B. `ZIP Code`
- kategorial ordinal: `Education`
- kategorial binär: `Personal Loan`

## 7. Plausibilität und auffällige Werte prüfen

Unplausible Werte und auffällige Werte sind nicht dasselbe:

- **unplausibel:** fachlich nicht sinnvoll oder unmöglich → möglicher Datenfehler
- **auffällig:** ungewöhnlich, aber möglicherweise korrekt

> **Auffällig bedeutet nicht automatisch fehlerhaft. Plausibilität muss fachlich beurteilt werden.**


```python
# Wertebereiche ausgewählter Merkmale
df_bank[["Age", "Experience", "Income", "CCAvg", "Mortgage"]].agg(["min", "max"])
```

Beispiele:

- `Experience = -3` → fachlich nicht plausibel
- `Income = 8` → sehr niedrig, aber nicht automatisch fehlerhaft
- `Income = 224` → sehr hoch, aber ebenfalls nicht automatisch fehlerhaft
- `CCAvg = 10` → auffällig hoch und deshalb genauer zu betrachten
- `Mortgage = 0` → kein fehlender Wert; es ist kein positiver Hypothekenwert erfasst


## 8. Metrische Merkmale untersuchen

Bei metrischen Merkmalen interessieren Lage, Streuung und Wertebereich. `describe()` liefert dafür zentrale deskriptive Kennzahlen.

```python
df_bank[["Age", "Income", "CCAvg", "Experience", "Mortgage"]].describe()
```

### Was verraten die Kennzahlen?

Erste Hinweise aus den Kennzahlen:

- `Experience`: Minimum von −3 ist fachlich nicht plausibel.
- `Income`: Mittelwert liegt über dem Median; Maximum liegt weit oberhalb des 3. Quartils.
- `CCAvg`: ebenfalls Mittelwert > Median und großer Abstand zwischen 3. Quartil und Maximum.
- `Mortgage`: Median = 0, aber deutlich höheres Maximum.

Kennzahlen geben **erste Hinweise auf Auffälligkeiten**. Ob ein Wert fehlerhaft, ungewöhnlich oder durch die Verteilung erklärbar ist, muss anschließend genauer geprüft werden.

## 9. Verteilungen metrischer Merkmale: Histogramm

### Beispiel: `Income`

Aus der Statistik ist das Histogramm als Dichtediagramm bekannt:

- Beobachtungen werden zu Klassen zusammengefasst.
- Die **Fläche** eines Balkens entspricht der relativen Häufigkeit der Beobachtungen in dieser Klasse.
- Die Gesamtfläche beträgt 1.
- Bei unterschiedlichen Klassenbreiten ist die Balkenhöhe eine **Dichte**, nicht unmittelbar eine Häufigkeit.

Seaborn verwendet bei `histplot()` standardmäßig `stat="count"`. Mit `stat="density"` entsteht die Dichtedarstellung.

`kde=True` ergänzt eine **Kernel Density Estimate (KDE)**: glättet die beobachteten Daten und approximiert die Form einer zugrunde liegenden stetigen Verteilung.

```python
sns.histplot(
    data=df_bank,
    x="Income",
    bins=25,
    stat="density",
    kde=True
)

plt.title("Verteilung von Income")
plt.show()
```

### Interpretation

- Die Verteilung ist deutlich **rechtsschief**.
- Ein großer Teil der Kunden liegt im unteren und mittleren Einkommensbereich.
- Nach rechts reicht die Verteilung bis zu deutlich höheren Einkommen.
- Mittelwert > Median passt zur beobachteten Rechtsschiefe.
- Hohe Werte sind nicht automatisch fehlerhaft.

## 10. Boxplot: Verteilungen kompakt beschreiben

### Beispiel: `Income`

Ein Boxplot fasst Lage, Streuung und potenzielle Ausreißer kompakt zusammen.

- Die Box enthält die mittleren 50 % der Beobachtungen.
- Der Median liegt innerhalb der Box.
- Die Whisker reichen zu den äußersten beobachteten Werten innerhalb der 1,5-IQR-Grenzen.
- Punkte außerhalb der Whisker werden als potenzielle Ausreißer markiert.

**Wichtig:** Ein Punkt außerhalb der Whisker ist nicht automatisch ein fehlerhafter Wert.

```python
sns.boxplot(
    data=df_bank,
    y="Income"
)

plt.title("Boxplot von Income")
plt.show()
```

Histogramm und Boxplot ergänzen sich:

- Das Histogramm zeigt die **Verteilungsform**.
- Der Boxplot zeigt **Lage, Streuung und potenzielle Ausreißer**.

Beim Einkommen unterstützen langer oberer Whisker und hohe potenzielle Ausreißer den bereits aus dem Histogramm gewonnenen Hinweis auf eine rechtsschiefe Verteilung.

## 11. Zusammenhänge zwischen Merkmalen untersuchen

Die EDA betrachtet nicht nur einzelne Merkmale, sondern auch Beziehungen zwischen Merkmalen und mögliche Unterschiede zwischen den Kundengruppen.

Ein Scatterplot stellt zwei metrische Merkmale gemeinsam dar. Die Farbe kann zusätzlich die Zielvariable `Personal Loan` kennzeichnen.

Gesucht werden Hinweise auf:

- positive oder negative Zusammenhänge
- unterschiedliche Gruppenstrukturen
- besondere Verteilungsmuster
- auffällige Beobachtungen

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4))

sns.scatterplot(
    data=df_bank,
    x="Age",
    y="Income",
    hue="Personal Loan",
    ax=axes[0],
    legend=False
)
axes[0].set_title("Age und Income")

sns.scatterplot(
    data=df_bank,
    x="Income",
    y="CCAvg",
    hue="Personal Loan",
    ax=axes[1],
    legend=False
)
axes[1].set_title("Income und CCAvg")

sns.scatterplot(
    data=df_bank,
    x="CCAvg",
    y="Mortgage",
    hue="Personal Loan",
    ax=axes[2]
)
axes[2].set_title("CCAvg und Mortgage")

plt.tight_layout()
plt.show()
```

### Erste Beobachtungen

- **Age und Income:** kein ausgeprägter Zusammenhang erkennbar; starke Überlagerung der Kundengruppen.
- **Income und CCAvg:** deutlicher positiver Zusammenhang; `Personal Loan = 1` tritt häufiger bei höheren Werten auf.
- **CCAvg und Mortgage:** kein klarer linearer Zusammenhang; auffällig ist die starke Häufung bei `Mortgage = 0`.

Die horizontale Struktur bei `Mortgage = 0` ist kein Ausreißerproblem, sondern ein besonderes Verteilungsmuster: Bei vielen Kunden ist kein positiver Hypothekenwert erfasst.

## 12. Korrelationen zwischen ausgewählten Merkmalen

Die Korrelationsmatrix ergänzt den visuellen Überblick quantitativ. Der Pearson-Korrelationskoeffizient liegt zwischen −1 und +1 und beschreibt **lineare** Zusammenhänge.

Bei einer mit 0/1 codierten binären Variable wie `Personal Loan` entspricht die Pearson-Korrelation mit einer metrischen Variable der punktbiserialen Korrelation. Sie kann hier als explorativer Hinweis auf einen Zusammenhang genutzt werden.

```python
correlation_variables = [
    "Age",
    "Income",
    "CCAvg",
    "Mortgage",
    "Personal Loan"
]

correlation_matrix = df_bank[correlation_variables].corr()
correlation_matrix
```

```python
plt.figure(figsize=(7, 5))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f"
)

plt.title("Korrelation ausgewählter Merkmale")
plt.tight_layout()
plt.show()
```

### Für die Analysefrage besonders interessant

- `Income` weist unter den betrachteten metrischen Merkmalen die stärkste positive Korrelation mit `Personal Loan` auf.
- `CCAvg` zeigt ebenfalls einen erkennbaren Zusammenhang.
- `Age` und `Mortgage` zeigen deutlich schwächere Zusammenhänge.

Die Heatmap dient als **Screening**. Korrelation allein beweist weder Kausalität noch endgültige Modellrelevanz.

### Welche Merkmale genauer untersuchen?

## 13. `Income` nach `Personal Loan`: Verteilungen vergleichen

Da die Gruppen mit 4.520 zu 480 Kunden sehr unterschiedlich groß sind, werden die Dichten **getrennt normiert**. `common_norm=False` verhindert, dass die größere Gruppe allein aufgrund ihrer Größe die Darstellung dominiert.

```python
sns.histplot(
    data=df_bank,
    x="Income",
    hue="Personal Loan",
    bins=25,
    stat="density",
    common_norm=False,
    element="step",
    kde=True
)

plt.title("Income nach Personal Loan")
plt.show()
```

Beobachtungen:

- `Personal Loan = 1`: Verteilung deutlich zu höheren Einkommen verschoben.
- `Personal Loan = 0`: Einkommen häufiger im niedrigeren und mittleren Bereich.
- Beide Verteilungen überlappen.

`Income` unterscheidet die beiden Kundengruppen sichtbar, trennt sie aber nicht vollständig.

```python
sns.boxplot(
    data=df_bank,
    x="Personal Loan",
    y="Income"
)

plt.title("Income nach Personal Loan")
plt.show()
```

Der Boxplot bestätigt den im Histogramm sichtbaren Gruppenunterschied:

- Median und Quartile liegen bei `Personal Loan = 1` höher.
- Die Wertebereiche beider Gruppen überlappen.
- Auch innerhalb beider Gruppen treten hohe Einkommen auf.

### `CCAvg` nach `Personal Loan` vergleichen

Auch `CCAvg` hat sich in den bisherigen Analysen als interessanter Kandidat gezeigt. Verteilung und Boxplot zeigen den Gruppenunterschied aus zwei Perspektiven.


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

sns.histplot(
    data=df_bank,
    x="CCAvg",
    hue="Personal Loan",
    bins=25,
    stat="density",
    common_norm=False,
    element="step",
    kde=True,
    ax=axes[0]
)
axes[0].set_title("Verteilung von CCAvg")

sns.boxplot(
    data=df_bank,
    x="Personal Loan",
    y="CCAvg",
    ax=axes[1]
)
axes[1].set_title("CCAvg nach Personal Loan")

plt.tight_layout()
plt.show()
```

Beobachtungen:

- Kunden mit `Personal Loan = 1` weisen tendenziell höhere Kreditkartenausgaben auf.
- Die Verteilung ist zu höheren Werten verschoben.
- Median und Quartile liegen bei `Personal Loan = 1` höher.
- Die beiden Gruppen überlappen deutlich.

Auch `CCAvg` unterscheidet die Kundengruppen, ermöglicht aber keine eindeutige Trennung.

## 14. Pairplot: mehrere Merkmale systematisch vergleichen

Zum Abschluss der explorativen Zusammenhangsanalyse fasst ein Pairplot mehrere numerische Merkmale gemeinsam zusammen:

- außerhalb der Diagonalen: Scatterplots
- auf der Diagonalen: Verteilungen der einzelnen Merkmale
- `hue="Personal Loan"`: Unterscheidung der Kundengruppen
- `corner=True`: spiegelbildlich doppelte Scatterplots werden ausgeblendet

Der Pairplot dient hier als **Screening und Zusammenfassung** der zuvor einzeln betrachteten Muster.


```python
sns.pairplot(
    data=df_bank,
    vars=["Age", "Income", "CCAvg", "Mortgage"],
    hue="Personal Loan",
    corner=True,
    diag_kind="hist"
)

plt.show()
```

### Was zeigt der Pairplot?

- `Age`: starke Überlagerung der beiden Kundengruppen.
- `Income`: `Personal Loan = 1` tritt häufiger bei höheren Einkommen auf.
- `CCAvg`: höhere Kreditkartenausgaben treten bei `Personal Loan = 1` häufiger auf.
- `Income` und `CCAvg`: positiver Zusammenhang zwischen beiden Merkmalen.
- `Mortgage`: auffällige Häufung bei 0; ansonsten starke Überlagerung der Gruppen.

Der Pairplot liefert **explorative Hinweise**. Er beweist weder statistische Signifikanz noch kausale Zusammenhänge.
